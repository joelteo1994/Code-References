### Cheatsheet for brew vs conda vs pip 

In Python and the data science ecosystem, you are likely to encounter three major tools for installing software and managing environments: Homebrew, Conda, and Pip. Whilst they may seem to overlap at first, each serves a different and complementary purpose, and choosing the right one can make the development workflow smoother. 
- Homebrew: macOS package manager that installs software at the system level; everything from command-line tools like git, wget and ffmpeg to full programming languages like python@3.11. 
    - It is not Python-specific, rather it is like apt on Ubuntu or yum on CentOS 
    - Use this if installing something that your system needs to access globally (especially CLI tool), which ensures that the tools are available in your terminal, independent of Python environments 
- Conda: All-in-One Environment and Package Manager that allows creation of isolated environments that contain specific versions of Python and all the libraries and dependencies you need. Also supports installing non-Python binaries like libjpeg, cuda, or graphviz, making it especially powerful for ML workflows. 
    - Use conda if working with scientific computing, machine learning, or complex environments that need both Python and system dependices 
    - With conda, you can: 
        - Create a clean, sandboxed environment for every project 
        - Use different Python versions across projects 
        - Avoid dependency hell by building powerful resolution logic
- Pip: Python's default package installer and connects to the Python Package Index (PyPI) to install libraries. It is lean, fast and widely used, but only handles Python packages. Has no built-in support for system dependencies or Python version management 
    - Works best inside an existing environment - either one created by conda or by Python's own venv
    - Use pip when just need to install or upgrade Python packages like transformers, fastapi, or scikit-learn

Overall mental model 
- Homebrew to prepare the system 
    - e.g. brew install git python@3.11 awscli
- Conda to prepare isolated, fully sandboxed (project-specific) environments
    - conda create -n myproject python=3.10
        - -n is short for --name, tells conda to create new environment and name it myproject 
        - replace myproject with your desired name 
        - general form: conda create -n <your_env_name> python=<version> [other packages...]
            - e.g. conda create -n sec_nlp_env python=3.9 pandas numpy
    - conda activate myproject
    - conda info to see more details about current active environment 
    - conda env list to list all environments, and the currently active one will have an asterisk (*)
- Conda and Pip to fill that environment with Python packages 
    - Can use pip after installing as many packages as possible via conda due to conda's stronger dependency pathway resolver and ability to pull compatible prebuilt binaries (especially from conda-forge)
    - But if used both conda and pip, export the environment like this conda env export > environment.yml (the only customisable part is environment.yml; replace with your desired output file name e.g. my_env_snapshot.yml)
        - In environment.yml, pip packages are listed like this 
            - dependencies: 
                - python=3.10
                - pandas 
                - pip 
                - pip: 
                    - fastapi
                    - transformers 
        - Note: yml is a configuration file format written in YAML (YAML Ain't Markup Language), which is a more readable, cleaner cousin of JSON. Used to define packages, dockers (multiple containers) etc
            - recall JSON (JavaScript Object Notation) is very much like a Python dictionary, a lightweight, text-based format for storing and exchanging data. This is the main format that lets different systems (e.g. browsers, servers, apps) talk to each other. Key-value format that allows lists (to allow more than one value per key) and nested structures (value can be lists or dictionaries or even lists of dictionaries). 
    - Then recreate it anywhere with conda env create -f environment.yml (again, the only customisable part is environment.yml which you should insert your requirements file). 
        - -f flag is short for --file, it tells conda to read the environment config from a file instead of expecting you to type the full list of packages manually. Hence, you can just pass in environment.yml which contains everything Conda needs to create the environment. 

#### Summary Table 

| Feature / Tool      | **Homebrew** (`brew`)                      | **Conda** (`conda`)                              | **Pip** (`pip`)                                |
|---------------------|--------------------------------------------|--------------------------------------------------|-------------------------------------------------|
| **Primary Purpose** | Install system-wide CLI tools, libraries   | Install Python environments + packages           | Install Python packages (env-based or global)   |
| **Search Command**  | `brew search <name>`                       | `conda search <name>`                            | `pip search <name>` (deprecated)                |
| **Install Command** | `brew install <pkg>`                       | `conda install <pkg>`                            | `pip install <pkg>`                             |
| **Version Control** | `brew install pkg@<version>`               | `conda install pkg=1.2.3`                        | `pip install pkg==1.2.3`                         |
| **Where It Installs** | System-wide (`/opt/homebrew`, `/usr/local`) | Per environment (conda envs)                   | Per environment (venv or conda) or globally     |
| **Package Sources** | Homebrew core + taps (GitHub-based)        | Conda channels (e.g., `defaults`, `conda-forge`) | PyPI (Python Package Index)                     |
| **Handles Non-Python?** | ✅ Yes (Node, Git, Tesseract, etc.)         | ✅ Yes (via `conda-forge`)                       | ❌ No — Python-only                              |
| **Dependency Mgmt** | Handles deps for system packages           | Handles all deps within env                      | Minimal dep resolution; may break environments  |
| **Virtual Envs?**   | ❌ No virtual environments                  | ✅ Native support (`conda create -n env`)         | ✅ Needs `venv`, `virtualenv`, or `conda`        |
| **Uninstall**       | `brew uninstall <pkg>`                     | `conda remove <pkg>`                             | `pip uninstall <pkg>`                           |
| **Update**          | `brew upgrade <pkg>`                       | `conda update <pkg>`                             | `pip install --upgrade <pkg>`                   |


### Difference between virtual environment and Jupyter Kernel 

#### 📘 1. What's the Difference?

| Concept | `conda` / `venv` (Environment)       | `Jupyter Kernel`                          |
|---------|--------------------------------------|-------------------------------------------|
| What it is | A sandboxed Python interpreter with isolated packages | A runtime that Jupyter uses to execute code |
| How it's created | `conda create -n env`, or `python -m venv env` | `ipykernel` installed inside an env and registered |
| Purpose | Isolate dependencies per project | Select and run code in Jupyter using a specific env (i.e. tells Jupyter which interpreter to run code with) |
| Visibility | Not visible in Jupyter unless manually registered | Appears in the Jupyter Kernel dropdown, but must be explicitly registered as per below steps |
| Link | Kernel *points to* an environment | Can't run unless the env is set up and active |
| Link | Lives at ~/miniconda3/envs/myenv | Kernel config lives at ~/.local/share/jupyter/kernels/myenv/ |
---

#### 🧩 2. Manual Steps to Register an Env as Jupyter Kernel

```bash
# Step 1: Activate your environment
conda activate myenv  # or source myenv/bin/activate if using venv

# Step 2: Install ipykernel (if not already)
pip install ipykernel

# Step 3: Register the environment as a Jupyter kernel
python -m ipykernel install --user --name=myenv --display-name "Python (myenv)"

# Step 4: List all Jupyter kernels with 
jupyter kernelspec list 

# Step 5: To remove a kernel: 
jupyter kernelspec uninstall myenv
