# 🔁 Python Loops Cheatsheet (For Data Science & AI)

---

## 1. 🧠 General Idea

Loops allow **repetition** in your code — iterating over datasets, tuning model parameters, or processing data batches.

In many data science and AI workflows, we often need to **repeat a set of operations multiple times**, either:

- For a **fixed number of iterations** (e.g. looping over 100 training epochs),
- Or **until a certain condition is met** (e.g. until loss drops below a threshold).

Rather than writing the same block of code again and again, loops let us **automate this repetition** cleanly and efficiently.

Typical loop use cases include:
- **Traversing collections** like lists, arrays, tensors, or DataFrames
- **Running simulations** until a convergence criterion is met
- **Processing batches of data** in model training
- **Performing grid searches** or hyperparameter sweeps
- **Generating or transforming features** using a consistent logic

By using loops, we avoid code duplication and gain control over execution flow, making it easier to write scalable and maintainable pipelines.

---

## 2. 📌 When to Use Which Loop

Different loop structures are optimized for different scenarios. Choosing the right type depends on:

- Whether the number of iterations is **known or unknown** in advance
- Whether you're working with **iterables** (like lists, arrays, or DataFrames)
- Whether you're **transforming data**, **traversing structures**, or **checking conditions**
- Whether your goal is **clarity**, **conciseness**, or **control**

Here's a breakdown of the most commonly used loop constructs in Python for data science and AI tasks:

| Loop Type         | Use Case                                                                 | Common With                          |
|------------------|--------------------------------------------------------------------------|--------------------------------------|
| `for` loop        | Known number of steps or iterating through a collection                 | Lists, arrays, DataFrames            |
| `while` loop      | Unknown number of steps; continue while condition is true               | Convergence checks, simulations      |
| `enumerate`       | Need index + value from iterable                                        | Feature selection, indexed traversal |
| `zip`             | Iterate over two or more collections together                           | Predicted vs actual, config pairing  |
| List Comprehension| Concise transformation of collections                                   | Feature cleaning, filtering          |
| `break/continue`  | Exit or skip parts of loop under condition                              | Early stopping, selective logic      |
| Nested Loops      | Iterate through combinations (e.g. grid search, 2D arrays)              | Model tuning, matrix ops             |

---

### 🧠 Additional Notes

- **Use `for` loops** when you have a collection of items and want to perform an operation for each item. They're ideal for cleanly iterating through rows in a DataFrame, lines in a file, or samples in a dataset.

- **Use `while` loops** when the loop termination depends on a **condition that evolves** within the loop — for example, waiting for a loss value to drop below a threshold, or retrying a model training until a desired accuracy is reached.

- **Use `enumerate`** when you need both the item and its index — this saves you from writing extra lines like `for i in range(len(...))`.

- **Use `zip`** when working with two or more iterables at once — for example, comparing predicted vs. actual values, or zipping model parameters with their corresponding gradients.

- **Use list comprehensions** when you want to create a **new list (or Series, or column)** by applying a transformation to every item in a collection. They're concise, readable, and often faster than `for` loops. For example:
  
  ```python
  # Standard loop
  cleaned = []
  for text in raw_texts:
      cleaned.append(text.strip().lower())

  # List comprehension
  cleaned = [text.strip().lower() for text in raw_texts]

## 3. 🔣 Abstract Syntax

This section shows how each type of loop is structured. Use these patterns as mental templates when writing your own.

---

### `for` loop — loop over an iterable (e.g. list, array, range)
```python
for item in iterable:
    # do something with item
```

🔹 Common variant using `range()`:
```python
for i in range(5):
    print(i)
```

---

### `while` loop — repeat while a condition remains True
```python
while condition:
    # do something until condition becomes False
```

---

### `enumerate` — loop with both index and value
```python
for i, val in enumerate(iterable):
    # i = index, val = element at i
```

---

### `zip` — iterate over multiple collections in parallel
```python
for a, b in zip(list1, list2):
    # do something with both a and b
```

---

### List Comprehension — compact way to transform or filter a list
```python
[expression for item in iterable if condition]
```

🔹 Example:
```python
squares = [x**2 for x in range(10) if x % 2 == 0]
```

---

### Nested Loop — loop within a loop (e.g. for grids or pairwise comparisons)
```python
for x in outer:
    for y in inner:
        # do something with x and y
```

---

### `for` + `else` — only runs `else` block if loop completes without `break`
```python
for x in iterable:
    if some_condition(x):
        break  # loop exited early
else:
    # runs only if loop did NOT break
    print("No break encountered")
```

🔹 Useful for: confirming no match found, fallbacks after search

## 4. 🧪 Deep Examples for Data Science & AI

This section shows how loop constructs apply in real-world data science and machine learning scenarios — from feature processing to training loops.

---

### ✅ DataFrame Row Iteration (via `iterrows`)
Useful when you need to access individual row values for logic or custom operations.
```python
import pandas as pd
df = pd.DataFrame({'feature': [1, 2, 3], 'label': [0, 1, 0]})

for idx, row in df.iterrows():
    print(f"Feature: {row['feature']}, Label: {row['label']}")
```

---

### ✅ Manual Training Loop (PyTorch-style)
Classic nested loop: outer loop for epochs, inner for batches.
```python
for epoch in range(epochs):
    for batch in dataloader:
        X, y = batch
        y_pred = model(X)
        loss = loss_fn(y_pred, y)
        loss.backward()
        optimizer.step()
        optimizer.zero_grad()
```

---

### ✅ Convergence Loop with `while`
Useful in optimization or simulations where you don’t know how many steps you'll need.
```python
loss = float('inf')
iteration = 0

while loss > 0.01 and iteration < 1000:
    y_pred = model(X)
    loss = loss_fn(y_pred, y)
    optimizer.step()
    iteration += 1
```

---

### ✅ List Comprehension for Feature Cleaning
A concise and fast way to apply transformations to all elements in a column.
```python
df['cleaned'] = [x.strip().lower() for x in df['raw_text']]
```

---

### ✅ `zip` for Side-by-Side Comparison
Align actual and predicted values for reporting, evaluation, or visualization.
```python
actual = [1, 0, 1]
pred = [1, 1, 0]

for a, p in zip(actual, pred):
    print(f"True: {a}, Pred: {p}")
```

---

### ✅ `enumerate` for Indexed Access
Track the index along with values — useful in metrics logging or debugging.
```python
for idx, score in enumerate(scores):
    print(f"Score {idx} = {score}")
```

---

### ✅ Nested Loop for Grid Search
Try all combinations of hyperparameters — foundational for basic model tuning.
```python
learning_rates = [0.01, 0.1]
batch_sizes = [16, 32]

for lr in learning_rates:
    for bs in batch_sizes:
        print(f"Trying model with lr={lr}, batch_size={bs}")
```

### ✅ `for` + `else`: Loop Completion Check
Rare but useful for confirming an exhaustive search without early exit.
```python
for x in [1, 2, 3]:
    if x == 4:
        break
else:
    print("Completed without hitting break!")  # This runs
```

## 5. 🚀 Pro Tips for Looping in Data Science

- 🧮 **Prefer vectorization** (NumPy, pandas) over manual loops for large data.
- ⚡ Use **list comprehensions** for speed and readability.
- 🧰 Use `tqdm` to show progress in long-running loops:
```python
from tqdm import tqdm
for i in tqdm(range(100)):
    train_step(i)
```
- ⚠️ Avoid modifying lists/dicts during iteration — use `.copy()` if needed.
- 🧠 Use loops + early stopping for convergence checks or evaluation thresholds.
